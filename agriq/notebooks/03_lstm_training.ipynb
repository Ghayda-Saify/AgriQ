# 03 — LSTM Price Forecasting

Train LSTM on FAO price history and visualize forecasts.

```python
import sys
sys.path.insert(0, "..")

from pathlib import Path
from src.models.train_lstm import train_lstm, forecast_prices
import pandas as pd
import numpy as np

prices_path = Path("../data/raw/fao_prices.csv")
model_path = Path("../models/lstm_price.pth")
model = train_lstm(prices_path, model_path, crop="Wheat")

df = pd.read_csv(prices_path)
history = df.loc[df["item"] == "Wheat", "price_usd_per_tonne"].values.astype(np.float32)
forecasts = forecast_prices(model, history, seq_len=3, horizon=3)
print("Forecasts:", forecasts)
```
