# 01 — Data Exploration

Explore raw datasets: Kaggle crop recommendation, FAO prices, and ARIJ yields.

```python
import pandas as pd
from pathlib import Path

DATA = Path("../data/raw")
kaggle = pd.read_csv(DATA / "kaggle_crop_recommendation.csv")
fao = pd.read_csv(DATA / "fao_prices.csv")
arij = pd.read_csv(DATA / "arij_yields.csv")

print("Kaggle shape:", kaggle.shape)
print("FAO shape:", fao.shape)
print("ARIJ shape:", arij.shape)
kaggle.head()
```

```python
import matplotlib.pyplot as plt

kaggle["label"].value_counts().plot(kind="bar", title="Crop distribution")
plt.show()
```
